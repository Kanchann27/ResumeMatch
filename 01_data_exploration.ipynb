```python
def clean_text(text):
    tokens = word_tokenize(text)

    tokens = [
        word for word in tokens
        if word.lower() not in english_stopwords
    ]

    tokens = [
        word for word in tokens
        if word not in string.punctuation
    ]

    tokens = [
        word.lower()
        for word in tokens
    ]

    return " ".join(tokens)
```

```python
clean_resume_text = clean_text(resume)
clean_job_text = clean_text(job_description)

print("CLEAN RESUME:")
print(clean_resume_text)

print("\nCLEAN JOB DESCRIPTION:")
print(clean_job_text)
```

```python
def extract_skills(text, skills):
    found_skills = []

    for skill in skills:
        if skill in text:
            found_skills.append(skill)

    return found_skills
```

```python
resume_skills = extract_skills(clean_resume_text, skills)
job_skills = extract_skills(clean_job_text, skills)

print("Resume Skills:")
print(resume_skills)

print("\nJob Description Skills:")
print(job_skills)
```

```python
def compare_skills(resume_skills, job_skills):
    matching = list(set(resume_skills) & set(job_skills))
    missing = list(set(job_skills) - set(resume_skills))
    extra = list(set(resume_skills) - set(job_skills))

    return matching, missing, extra
```

```python
matching_skills, missing_skills, extra_skills = compare_skills(
    resume_skills,
    job_skills
)

print("Matching Skills:")
print(matching_skills)

print("\nMissing Skills:")
print(missing_skills)

print("\nAdditional Resume Skills:")
print(extra_skills)
```

```python
def calculate_skill_match(matching_skills, job_skills):
    if len(job_skills) == 0:
        return 0

    percentage = (len(matching_skills) / len(job_skills)) * 100

    return percentage
```

```python
skill_match_percentage = calculate_skill_match(
    matching_skills,
    job_skills
)

print(f"Skill Match Percentage: {skill_match_percentage:.2f}%")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def calculate_text_similarity(resume_text, job_text):
    documents = [resume_text, job_text]

    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform(documents)

    similarity = cosine_similarity(
        tfidf_matrix[0],
        tfidf_matrix[1]
    )

    return similarity[0][0]
```

```python
similarity_score = calculate_text_similarity(
    clean_resume_text,
    clean_job_text
)

similarity_percentage = similarity_score * 100

print(f"Resume-Job Similarity: {similarity_percentage:.2f}%")
```

```python
def analyze_resume_job(resume_text, job_text, skills):
    # Clean both texts
    clean_resume = clean_text(resume_text)
    clean_job = clean_text(job_text)

    # Extract skills
    resume_skills = extract_skills(clean_resume, skills)
    job_skills = extract_skills(clean_job, skills)

    # Compare skills
    matching, missing, extra = compare_skills(
        resume_skills,
        job_skills
    )

    # Calculate skill match percentage
    skill_match = calculate_skill_match(
        matching,
        job_skills
    )

    # Calculate text similarity
    text_similarity = calculate_text_similarity(
        clean_resume,
        clean_job
    )

    return {
        "resume_skills": resume_skills,
        "job_skills": job_skills,
        "matching_skills": matching,
        "missing_skills": missing,
        "additional_skills": extra,
        "skill_match_percentage": skill_match,
        "text_similarity_percentage": round(text_similarity * 100, 2)
    }
```

```python
result = analyze_resume_job(
    resume,
    job_description,
    skills
)

print(result)
```

```python
print("RESUME - JOB ANALYSIS")
print("=" * 40)

print("Matching Skills:", ", ".join(result["matching_skills"]))
print("Missing Skills:", ", ".join(result["missing_skills"]))
print("Additional Skills:", ", ".join(result["additional_skills"]))
print(f"Skill Match: {result['skill_match_percentage']:.2f}%")
print(f"Text Similarity: {float(result['text_similarity_percentage']):.2f}%")
```
